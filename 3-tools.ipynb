```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
model=init_chat_model("groq:openai/gpt-oss-20b")
response=model.invoke("Hello, how are you?")
print(response.content)
```

```python
### Creating Tool // This the Second way to bind tools
### One way i already learned in file 1-langchaininto.ipynb

from langchain.tools import tool

@tool
def get_weather(location:str)->str:
    """Get the current weather in a given location."""
    return f"The current weather in {location} is sunny."


tool_call_function=model.bind_tools([get_weather])

response=tool_call_function.invoke("What is the weather of Jeddah")
# print(response)

for result in response.tool_calls:
    print(f"This is Tool: {result['name']}, And this is argument{result['args']}")
```
